```python
# Step 1
# Import libraries needed for web scraping
from bs4 import BeautifulSoup
import requests
import pandas as pd
print("✅ Step 1: Libraries imported!")
```

```python
# Step 2
#set URL of website in a variable
url = "https://www.scrapethissite.com/pages/forms/"
page = requests.get(url)

#Use BeautifulSoup to extract the HTML content
soup = BeautifulSoup(page.content, "html.parser")
print("✅ Step 2: Page fetched and parsed!")
```

```python
# Step 3: Extract the Table Data
#Extract the table with the Hockey Scores
hockey_table = soup.find("table", class_="table")

#Extract the column headings
table_titles = hockey_table.find_all("th")
columns = [title.text.strip() for title in table_titles]
print("✅ Step 3: Columns found:", columns)
```

```python
# Step 4: Store structured data into a Pandas DataFrame.
#Save the column headings onto a Pandas DataFrame
df = pd.DataFrame(columns=columns)
print("✅ Step 4: DataFrame created!")
```

```python
# Step 5: Export the final dataset to a .csv file.
#Extract the data row by row. First get all rows, then loop through each while stripping and saving data into the DataFrame
table_data = hockey_table.find_all('tr')[1:]  # Skip header row
for row in table_data:
    raw_data = row.find_all('td')
    each_raw_data = [data.text.strip() for data in raw_data]
#saving each row data as it is generated into the pandas data frame
    length = len(df)
    df.loc[length] = each_raw_data
print("✅ Step 5: All rows extracted!")
```

```python
# Step 5: Inspect results
#Inspect the resulting DataFrame
print(f"✅ Step 5: DataFrame shape: {df.shape}")
print(df.head())
print("\nLast 5 rows:")
print(df.tail())
```

```python
# Step 5: Export to CSV
#Save to a .csv file in the current folder
df.to_csv('./Hockey.csv', index=False)
print("✅ Step 5: Data saved to Hockey.csv!")
print("🎉 PROJECT COMPLETE!")
```
